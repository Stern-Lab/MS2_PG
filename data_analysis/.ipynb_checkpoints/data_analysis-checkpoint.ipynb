```python
def get_sumstats_per_line(df_for_sumstat_by_line_all_passages, passage=10, adaptive=''):
    """
    Transforms a mutations dataframe into a dictionary of lists containing 
    frequency-based summary statistics per line.
    df_for_sumstat_by_line -- df with the columns ['line', 'protein', 'syn_nonsyn', 'adaptive_moi10', 'freq']
    adaptive -- string, '' if regular definition of adaptive, 'weak_' for the weak definition
    """
    df_for_sumstat_by_line = df_for_sumstat_by_line_all_passages[df_for_sumstat_by_line_all_passages['passage']==passage]
    
    nonsyn_proteins = ['mat', 'cp', 'lys', 'rep']
    ada_col_name = adaptive+'adaptive_moi'
    adaptive_col = [col for col in df_for_sumstat_by_line.columns if col.startswith(ada_col_name)][0]
    
    
    lines = df_for_sumstat_by_line['line'].unique()
    all_sumstats = {}

    for line_id in lines:
        line_data = df_for_sumstat_by_line[df_for_sumstat_by_line['line'] == line_id]
        line_stat = []
        
        # Order: [protein-nonsyn-nonada, protein-nonsyn-ada]
        for prot in nonsyn_proteins:
            for is_adaptive in ['nonada', 'ada']:
                mask = (
                    (line_data['protein'] == prot) & 
                    (line_data['syn_nonsyn'] == 'nonsyn') & 
                    (line_data[adaptive_col] == is_adaptive)
                )
                line_stat.append(line_data.loc[mask, 'freq'].sum())

        # Order: [total-syn-nonada, total-syn-ada]
        for is_adaptive in ['nonada', 'ada']:
            mask = (
                (line_data['syn_nonsyn'] == 'syn') & 
                (line_data[adaptive_col] == is_adaptive)
            )
            line_stat.append(line_data.loc[mask, 'freq'].sum())
            
        all_sumstats[line_id] = torch.tensor(line_stat)
    all_sumstats['mean_sumstat'] = torch.stack(list(all_sumstats.values())).mean(dim=0)    
    return all_sumstats


def merge_dictionaries_with_lists(dict_list):
    merged_dict = {}
    for d in dict_list:
        for key, value in d.items():
            if key not in merged_dict:
                merged_dict[key] = []
            if isinstance(value, list):
                merged_dict[key].extend(value)
            else:
                merged_dict[key].append(value)
    return merged_dict

def merge_dictionaries_with_tensors(dict_list):
    merged_dict = {}
    for d in dict_list:
        for key, value in d.items():
            if key not in merged_dict:
                merged_dict[key] = []
            merged_dict[key].append(value)
    
    for key in merged_dict:
        merged_dict[key] = torch.cat(merged_dict[key])
        
    return merged_dict
```

```python
def get_genotype_sumstat(df, passage=10):
    max_muts = 9 # so 8 maximum muts
    new_index = [(mat, cp, lys, rep, ada, syn) for mat in range(max_muts) for cp in range(max_muts)
             for lys in range(max_muts) for rep in range(max_muts) for ada in range(max_muts)
             for syn in range(max_muts) if mat+cp+lys+rep+ada+syn<max_muts]
    
    df = df[df['passage']==passage]
    lines = df['line'].unique()
    all_sumstats = {}
    for line in lines:
        line_data = df[df['line'] == line]
        total_num_reads = len(line_data)
        grouped = line_data.groupby(['mat_nonsyn_nonada', 'cp_nonsyn_nonada', 'lys_nonsyn_nonada', 'rep_nonsyn_nonada','total_ada', 'total_syn_nonada']).size()/total_num_reads
        all_sumstats[line] = torch.Tensor(grouped.reindex(new_index).fillna(0).values.flatten())
    all_sumstats['mean_sumstat'] = torch.stack(list(all_sumstats.values())).mean(dim=0)
    return all_sumstats
```

```python
def get_full_geno_sumstat(df):
    binned_categories = [
        "mat_nonsyn_nonada","mat_nonsyn_ada", "cp_nonsyn_nonada","cp_nonsyn_ada", "lys_nonsyn_nonada", "lys_nonsyn_ada",
        "rep_nonsyn_nonada","rep_nonsyn_ada", "total_syn_nonada", "total_syn_ada"
    ]
    df_p10 = df[df['passage']==10].copy()
    # Apply Transformation: 0->0, 1-2->1, 3-4->2, 5-6->3, 7-8->4
    # Logic: (val + 1) // 2 for val > 0
    for col in binned_categories:
        new_col_name = f"{col}_binned"
        df_p10[new_col_name] = df_p10[col].apply(lambda x: (int(x) + 1) // 2 if x > 0 else 0)

    def generate_indexing(n_categories, max_total):
        # This generates all tuples (k1, k2, ..., kn) where sum(ki) <= max_total
        # We use max_total as the number of 'stars' and n_categories as 'bars'
        # Adding a slack variable (n + 1) handles the "<" constraint
        for combo in combinations(range(n_categories + max_total), n_categories):
            res = []
            prev = -1
            for val in combo:
                res.append(val - prev - 1)
                prev = val
            yield tuple(res)

    new_index = list(generate_indexing(10, 6 - 1))
    
    lines = df_p10['line'].unique()
    all_sumstats = {}
    for line in lines:
        line_data = df_p10[df_p10['line'] == line]
        total_num_reads = len(line_data)
        grouped = line_data.groupby(['mat_nonsyn_nonada_binned','mat_nonsyn_ada_binned', 'cp_nonsyn_nonada_binned', 'cp_nonsyn_ada_binned',
                          'lys_nonsyn_nonada_binned','lys_nonsyn_ada_binned', 'rep_nonsyn_nonada_binned','rep_nonsyn_ada_binned',
                          'total_syn_nonada_binned', 'total_syn_ada_binned']).size()/total_num_reads
        all_sumstats[line] = torch.Tensor(grouped.reindex(new_index).fillna(0).values.flatten())
    all_sumstats['mean_sumstat'] = torch.stack(list(all_sumstats.values())).mean(dim=0)
    
    return all_sumstats
```

## plot syn vs nonsyn mutations

```python
def process_data_for_whole_genome_plot(freqs_with_metadata_df, moi_label):
    """
    Processes the raw frequency data frame to get the mean sum of 
    frequencies per passage, syn/nonsyn, and adaptive status, 
    and adds the MOI label.
    """
    adaptive_col = freqs_with_metadata_df.columns[freqs_with_metadata_df.columns.str.contains('adaptive')][0]
    
    # Sum freqs per passage, line, syn/nonsyn, and adaptive status
    freq_sum_by_p_l = freqs_with_metadata_df.groupby(['passage', 'line', 'syn_nonsyn', adaptive_col])['freq'].sum().reset_index()
    
    # Calculate the mean sum across all lines
    mean_freq_sum = freq_sum_by_p_l.groupby(['passage', 'syn_nonsyn', adaptive_col])['freq'].mean().reset_index()
    
    pivot = mean_freq_sum.pivot_table(
        index=['passage', 'syn_nonsyn'],
        columns=adaptive_col,
        values='freq',
        fill_value=0
    ).reset_index()

    pivot = pivot.rename(columns={False: 'nonada', True: 'ada'})
    pivot['moi'] = moi_label
    return pivot

def plot_freq_sum_whole_genome_comparison(freqs_with_metadata_df_moi10, freqs_with_metadata_df_moi01):
    '''
    Plots the whole genome frequency sum for MOI-10 and MOI-0.1 

    arguments:
    freqs_with_metadata_df_moi10: pandas df for MOI=10 data.
    freqs_with_metadata_df_moi01: pandas df for MOI=0.1 data.
    '''
    pivot_moi10 = process_data_for_whole_genome_plot(freqs_with_metadata_df_moi10, 'MOI-10')
    pivot_moi01 = process_data_for_whole_genome_plot(freqs_with_metadata_df_moi01, 'MOI-0.1')
    
    combined_pivot = pd.concat([pivot_moi10, pivot_moi01])
    
    fig, ax = plt.subplots(figsize=(10, 6))

    base_colors = {
        'syn': '#1f77b4',       # blue
        'nonsyn': '#d62728'     # red
    }
    light_colors = {
        'syn': '#aec7e8',       # light blue
        'nonsyn': '#ff9896'     # light red
    }
    
    hatch_patterns = {
        'MOI-10': '',       # Solid
        'MOI-0.1': '///'    # Hatched
    }
    
  
    bar_width = 0.2 
    total_group_width = 4 * bar_width  # Total width of the Syn/Nonsyn/MOI group (0.8)

    passages = sorted(combined_pivot['passage'].unique())
    n_passages = len(passages)
    
    x_centers = np.arange(n_passages) 
    
    x_start_offset = -total_group_width / 2 # -0.4

    legend_handles = {}

    # (syn_nonsyn, moi, offset_from_start)
    bar_order = [
        ('syn', 'MOI-10', 0 * bar_width),
        ('syn', 'MOI-0.1', 1 * bar_width),
        ('nonsyn', 'MOI-10', 2 * bar_width),
        ('nonsyn', 'MOI-0.1', 3 * bar_width)
    ]

    
    for sn, moi, rel_offset in bar_order:
        
        xpos = x_centers + x_start_offset + rel_offset
        
        subset = combined_pivot[
            (combined_pivot['syn_nonsyn'] == sn) & 
            (combined_pivot['moi'] == moi)
        ]
        
        passage_mask = np.isin(passages, subset['passage'].tolist())
        current_xpos = xpos[passage_mask]
        current_subset = subset.sort_values(by='passage')

        base_color = base_colors[sn]
        light_color = light_colors[sn]
        hatch = hatch_patterns[moi]
        
        ax.bar(
            current_xpos,
            current_subset['nonada'],
            width=bar_width,
            color=base_color,
            edgecolor='black',
            hatch=hatch,
            label='_nolegend_'
        )

        key = f'{moi}'
        if key not in legend_handles:
            legend_handles[key] = ax.bar(0, 0, width=0, color='gray', edgecolor='black', hatch=hatch, label=key)
        
        key_sn = f'{sn} non-adaptive'
        if key_sn not in legend_handles:
            legend_handles[key_sn] = ax.bar(0, 0, width=0, color=base_color, label=key_sn)


        ax.bar(
            current_xpos,
            current_subset['ada'],
            width=bar_width,
            bottom=current_subset['nonada'],
            color=light_color,
            edgecolor='black',
            hatch=hatch,
            label='_nolegend_'
        )
        
        key_ada = f'{sn} adaptive'
        if key_ada not in legend_handles:
            legend_handles[key_ada] = ax.bar(0, 0, width=0, color=light_color, label=key_ada)


    ax.set_xticks(x_centers)
    ax.set_xticklabels(passages)
    
    ax.set_xlabel("Passage")
    ax.set_ylabel("Mean Sum of Frequencies Across Lines")
    ax.set_ylim(0, combined_pivot[['nonada', 'ada']].sum(axis=1).max() * 1.1)
    
    ax.set_title('Comparison of Whole Genome Mutation Frequencies: MOI-10 vs MOI-0.1')
    ax.grid(axis='y', linestyle='--', alpha=0.7)
    
    moi_labels = ['MOI-10', 'MOI-0.1']
    moi_handles = [legend_handles[l] for l in moi_labels]

    sn_labels = [
        'syn non-adaptive', 'syn adaptive', 
        'nonsyn non-adaptive', 'nonsyn adaptive'
    ]
    sn_handles = [legend_handles[l] for l in sn_labels]
    
    leg1 = ax.legend(moi_handles, moi_labels, title="MOI", 
                     loc='upper right', bbox_to_anchor=(0.15, 1.0))

    # Syn/Nonsyn/Adaptive Legend
    ax.add_artist(leg1)
    ax.legend(sn_handles, sn_labels, title="Mutation Type", 
              loc='upper right', bbox_to_anchor=(0.25, 0.8))

    plt.tight_layout()
    plt.show()
```

```python
def process_data_for_protein_plot(freqs_with_metadata_df, moi_label,ada=''):
    """
    Processes the raw frequency data frame to get the mean sum of 
    frequencies per passage, syn/nonsyn, protein, and adaptive status,
    and adds the MOI label.
    
    ada -- str, '' if regular definition of adaptive, 'weak_' if weak definition
    """
    adaptive_name = ada+'adaptive'
    adaptive_col = freqs_with_metadata_df.columns[freqs_with_metadata_df.columns.str.contains(adaptive_name)][0]

    freq_sum_by_p_l = freqs_with_metadata_df.groupby(['passage','line','syn_nonsyn','protein',adaptive_col])['freq'].sum().reset_index()
    mean_freq_sum = freq_sum_by_p_l.groupby(['passage','syn_nonsyn','protein',adaptive_col])['freq'].mean().reset_index()

    pivot = mean_freq_sum.pivot_table(
        index=['protein', 'passage', 'syn_nonsyn'],
        columns=adaptive_col,
        values='freq',
        fill_value=0
    ).reset_index()

    pivot = pivot.rename(columns={False: 'nonada', True: 'ada'})
    pivot['moi'] = moi_label
    return pivot


def plot_freq_sum_by_protein_comparison(freqs_with_metadata_df_moi10, freqs_with_metadata_df_moi01):
    """
    Creates a grid of subplots (one for each protein) where bars for MOI-10 
    and MOI-0.1 are grouped side-by-side for comparison.
    """
    pivot_moi10 = process_data_for_protein_plot(freqs_with_metadata_df_moi10, 'MOI-10')
    pivot_moi01 = process_data_for_protein_plot(freqs_with_metadata_df_moi01, 'MOI-0.1')
    
    combined_pivot = pd.concat([pivot_moi10, pivot_moi01])

    base_colors = {'syn': '#1f77b4', 'nonsyn': '#d62728'}
    light_colors = {'syn': '#aec7e8', 'nonsyn': '#ff9896'}
    hatch_patterns = {
        'MOI-10': '',       # Solid
        'MOI-0.1': '///'    # Hatched
    }
    
    bar_width = 0.18 
    total_group_width = 4 * bar_width  
    
    passages = [3,7,10] # range(1, 11) 
    n_passages = len(passages)
    x_centers = np.arange(n_passages) 
    x_start_offset = -total_group_width / 2 

    proteins_dict = {'mat':1182, 'cp':393, 'lys':228, 'rep':1638}
    proteins = sorted(proteins_dict.keys())
    n_proteins = len(proteins)

    ncols = math.ceil(math.sqrt(n_proteins))
    nrows = math.ceil(n_proteins / ncols)

    fig, axes = plt.subplots(
        nrows=nrows, ncols=ncols, figsize=(6*ncols, 6*nrows), sharey=True
    )
    axes = axes.flatten()
    
    legend_handles = {}
    
    bar_order = [
        ('syn', 'MOI-10', 0 * bar_width),
        ('syn', 'MOI-0.1', 1 * bar_width),
        ('nonsyn', 'MOI-10', 2 * bar_width),
        ('nonsyn', 'MOI-0.1', 3 * bar_width)
    ]

    for i, protein in enumerate(proteins):
        if i >= len(axes): 
            break
            
        ax = axes[i]
        subset_prot = combined_pivot[combined_pivot['protein'] == protein]

        for sn, moi, rel_offset in bar_order:
            
            xpos = x_centers + x_start_offset + rel_offset
            
            subset = subset_prot[
                (subset_prot['syn_nonsyn'] == sn) & 
                (subset_prot['moi'] == moi)
            ]
            
            passage_mask = np.isin(passages, subset['passage'].tolist())
            current_xpos = xpos[passage_mask]
            current_subset = subset.sort_values(by='passage')
            
            base_color = base_colors[sn]
            light_color = light_colors[sn]
            hatch = hatch_patterns[moi]
            
            # Plot non-adaptive base
            ax.bar(
                current_xpos, current_subset['nonada'], width=bar_width, color=base_color, 
                edgecolor='black', hatch=hatch, label='_nolegend_'
            )
            
            key_moi = f'{moi}'
            if key_moi not in legend_handles:
                legend_handles[key_moi] = ax.bar(0, 0, width=0, color='gray', edgecolor='black', hatch=hatch, label=key_moi)
            
            key_sn = f'{sn} non-adaptive'
            if key_sn not in legend_handles:
                legend_handles[key_sn] = ax.bar(0, 0, width=0, color=base_color, label=key_sn)

            # Stack adaptive portion on top
            ax.bar(
                current_xpos, current_subset['ada'], width=bar_width, bottom=current_subset['nonada'], 
                color=light_color, edgecolor='black', hatch=hatch, label='_nolegend_'
            )
            
            key_ada = f'{sn} adaptive'
            if key_ada not in legend_handles:
                legend_handles[key_ada] = ax.bar(0, 0, width=0, color=light_color, label=key_ada)

            
        ax.set_xticks(x_centers) 
        ax.set_xticklabels(passages)
        ax.set_xlabel("Passage",fontsize=14)
        ax.set_title(f"Protein: {protein}, Gene length: {proteins_dict[protein]}")
        ax.grid(axis='y', linestyle='--', alpha=0.7)


    for ax in axes[n_proteins:]:
        ax.axis("off")

    axes[0].set_ylabel("Mean Sum of Frequencies Across Lines",fontsize=14)
    axes[2].set_ylabel("Mean Sum of Frequencies Across Lines",fontsize=14)
    
    moi_labels = ['MOI-10', 'MOI-0.1']
    moi_handles = [legend_handles[l] for l in moi_labels]

    sn_labels = [
        'syn non-adaptive', f'syn adaptive', 
        'nonsyn non-adaptive', f'nonsyn adaptive'
    ]
    sn_handles = [legend_handles[l] for l in sn_labels]
    
    leg1 = fig.legend(moi_handles, moi_labels, title="MOI", 
                      loc='upper right', bbox_to_anchor=(0.18, 0.9))

    fig.add_artist(leg1)
    fig.legend(sn_handles, sn_labels, title="Mutation Type", 
               loc='upper right', bbox_to_anchor=(0.24, 0.82))


    fig.suptitle('Comparison of Protein Mutation Frequencies (MOI-10 vs MOI-0.1)', fontsize=16, y=1.00)
    plt.tight_layout(rect=[0, 0, 1, 0.98]) 
    plt.show()
```

```python
def process_data_for_line_plot(freqs_df, ada=''):
    """
    Processes the raw frequency data to get the sum of frequencies per 
    passage, line, syn/nonsyn, and protein.
    """
    adaptive_name = ada + 'adaptive'
    adaptive_col = freqs_df.columns[freqs_df.columns.str.contains(adaptive_name)][0]

    freq_sum = freqs_df.groupby(['passage', 'line', 'syn_nonsyn', 'protein', adaptive_col])['freq'].sum().reset_index()

    pivot = freq_sum.pivot_table(
        index=['protein', 'passage', 'line', 'syn_nonsyn'],
        columns=adaptive_col,
        values='freq',
        fill_value=0
    ).reset_index()

    pivot = pivot.rename(columns={False: 'nonada', True: 'ada'})
    return pivot

def plot_line_comparison_by_protein(freqs_with_metadata_df, moi_label):
    """
    Creates a grid of subplots (one for each protein) where bars for 
    different experimental lines are grouped for comparison.
    """
    pivot_df = process_data_for_line_plot(freqs_with_metadata_df)
    
    lines = sorted(pivot_df['line'].unique())
    passages = [3, 7, 10]
    proteins_dict = {'mat':1182, 'cp':393, 'lys':228, 'rep':1638}
    proteins = sorted(proteins_dict.keys())
    
    base_colors = {'syn': '#1f77b4', 'nonsyn': '#d62728'}
    light_colors = {'syn': '#aec7e8', 'nonsyn': '#ff9896'}
    line_hatches = ['', '///', '...', '\\\\', 'xxx', 'OO'] 
    hatch_dict = {line: line_hatches[i % len(line_hatches)] for i, line in enumerate(lines)}

    n_lines = len(lines)
    bar_width = 0.8 / (2 * n_lines) 
    total_group_width = n_lines * 2 * bar_width
    x_centers = np.arange(len(passages))
    x_start_offset = -total_group_width / 2

    fig, axes = plt.subplots(
        nrows=2, ncols=2, figsize=(14, 12), sharey=True
    )
    axes = axes.flatten()
    legend_handles = {}

    for i, protein in enumerate(proteins):
        ax = axes[i]
        subset_prot = pivot_df[pivot_df['protein'] == protein]

        for l_idx, line in enumerate(lines):
            for sn_idx, sn in enumerate(['syn', 'nonsyn']):
                
                rel_offset = (l_idx * 2 + sn_idx) * bar_width
                xpos = x_centers + x_start_offset + rel_offset
                
                subset = subset_prot[(subset_prot['line'] == line) & (subset_prot['syn_nonsyn'] == sn)]
                
                current_subset = pd.merge(pd.DataFrame({'passage': passages}), subset, on='passage', how='left').fillna(0)
                
                color = base_colors[sn]
                l_color = light_colors[sn]
                hatch = hatch_dict[line]

                # Plot non-adaptive
                ax.bar(xpos, current_subset['nonada'], width=bar_width, color=color, 
                       edgecolor='black', hatch=hatch)
                
                # Plot adaptive (stacked)
                ax.bar(xpos, current_subset['ada'], width=bar_width, bottom=current_subset['nonada'],
                       color=l_color, edgecolor='black', hatch=hatch)

                if f'Line {line}' not in legend_handles:
                    legend_handles[f'Line {line}'] = ax.bar(0, 0, width=0, color='white', edgecolor='black', hatch=hatch)
                if sn not in legend_handles:
                    legend_handles[sn] = ax.bar(0, 0, width=0, color=color)
                if f'{sn} adaptive' not in legend_handles:
                    legend_handles[f'{sn} adaptive'] = ax.bar(0, 0, width=0, color=l_color)

        ax.set_xticks(x_centers)
        ax.set_xticklabels(passages)
        ax.set_title(f"Protein: {protein} ({moi_label})")
        ax.set_ylabel("Sum of Frequencies")

    line_labels = [f'Line {l}' for l in lines]
    line_handles = [legend_handles[l] for l in line_labels]
    
    type_labels = ['syn', 'syn adaptive', 'nonsyn', 'nonsyn adaptive']
    type_handles = [legend_handles[t] for t in type_labels]

    leg1 = fig.legend(line_handles, line_labels, title="Experimental Lines", loc='center right', bbox_to_anchor=(1.1, 0.6))
    fig.add_artist(leg1)
    fig.legend(type_handles, type_labels, title="Mutation Status", loc='center right', bbox_to_anchor=(1.1, 0.4))

    plt.tight_layout(rect=[0, 0, 0.9, 1])
    plt.show()

```

## Mean frequencies comparison

```python
def create_summary_ferq_df(freqs_with_metadata_df, threshold, greater_than=True):
    moi = '10' if any(freqs_with_metadata_df.columns.str.contains('moi10')) else '0.1'
    freqs_with_metadata_df = freqs_with_metadata_df[freqs_with_metadata_df['passage']==10]
    if greater_than:
        g = '>'
        summary_freq_df = freqs_with_metadata_df[freqs_with_metadata_df['freq']>threshold]
    else:
        g = '<'
        summary_freq_df = freqs_with_metadata_df[freqs_with_metadata_df['freq']<=threshold]
    summary_freq_df = summary_freq_df.groupby(['line','protein']).agg(
        median_freq=('freq','median'),mean_freq=('freq','mean'),max_freq=('freq','max'), sum_freq=('freq','sum'))
    summary_freq_df['log10_median'] = np.log10(summary_freq_df['median_freq'])
    summary_freq_df['log10_mean'] = np.log10(summary_freq_df['mean_freq'])
    summary_freq_df.name = f'MOI-{moi}_summary_freq_df, freq {g} {threshold}'
    
    return summary_freq_df


def plot_moi_protein_grid_with_tests(
    df10: pd.DataFrame,
    df01: pd.DataFrame,
    metric: str = "mean",     # "median" or "mean" or "sum"
    use_log: bool = True,       # if True uses 'log10_median'/'log10_mean'; else 'median_freq'/'mean_freq', if metric=sum use_log should be False
    freq_threshold = '< 0.01',  # for title only
    title_additions = '',
    proteins = ("mat", "cp", "lys", "rep"),
    figsize=(9, 6),
    moi_labels=("0.1", "10"),
    point_size=70,
    line_alpha=0.5
):
    """
    Plot per-line mean/median mutation frequencies for MOI 0.1 vs 10, faceted by protein,
    and perform Welch's t-tests for difference in mean values between MOIs per protein.
    """
    
    metric = metric.lower()
    if metric not in {"mean", "median", "sum"}:
        raise ValueError("metric must be 'mean', 'median' or 'sum'")
    col = f"log10_{metric}" if use_log else f"{metric}_freq"

    def _prep(df, moi_value):
        tmp = df.reset_index()
        tmp.columns = ["line", "protein"] + list(tmp.columns[2:])
        tmp["MOI"] = moi_value
        return tmp[["line", "protein", "MOI", col]]

    t10 = _prep(df10, moi_labels[1])
    t01 = _prep(df01, moi_labels[0])
    all_df = pd.concat([t01, t10], ignore_index=True)
    all_df["line"] = all_df["line"].astype(str)
    all_df = all_df[all_df["protein"].isin(proteins)]
    all_df["protein"] = pd.Categorical(all_df["protein"], categories=proteins, ordered=True)

    unique_lines = all_df["line"].astype(str).unique()
    cmap = plt.get_cmap("tab10") if len(unique_lines) <= 10 else plt.get_cmap("tab20")
    color_map = {ln: cmap(i % cmap.N) for i, ln in enumerate(sorted(unique_lines))}

    ys = all_df[col].dropna().values
    ypad = 0.05 * (np.nanmax(ys) - np.nanmin(ys))
    y_min, y_max = np.nanmin(ys) - ypad, np.nanmax(ys) + ypad

    fig, axes = plt.subplots(2, 2, figsize=figsize, sharey=True)
    axes = np.array(axes).flatten()

    x_positions = {moi_labels[0]: 0, moi_labels[1]: 1}
    test_results = []

    for ax, prot in zip(axes, proteins):
        sub = all_df[all_df["protein"] == prot]
        for ln, grp in sub.groupby("line"):
            xs, ys_line = [], []
            for moi in (moi_labels[0], moi_labels[1]):
                val = grp.loc[grp["MOI"] == moi, col]
                if len(val) > 0:
                    xs.append(x_positions[moi])
                    ys_line.append(float(val.iloc[0]))
                else:
                    xs.append(None)
                    ys_line.append(None)
            if None not in xs and None not in ys_line:
                ax.plot(xs, ys_line, "-", lw=1.2, color=color_map[ln], alpha=line_alpha)
            for x, y in zip(xs, ys_line):
                if x is not None and y is not None:
                    ax.scatter(x, y, s=point_size, color=color_map[ln], edgecolor="black", linewidth=0.5, zorder=3)

        # --- Statistical test ---
        vals_01 = sub.loc[sub["MOI"] == moi_labels[0], col].dropna()
        vals_10 = sub.loc[sub["MOI"] == moi_labels[1], col].dropna()
        if len(vals_01) >= 2 and len(vals_10) >= 2:
            if vals_01.std() < 1e-12:
                vals_01 = vals_01 + np.random.normal(0, 1e-12, len(vals_01))
            if vals_10.std() < 1e-12:
                vals_10 = vals_10 + np.random.normal(0, 1e-12, len(vals_10))
            # ONE-SIDED TEST: 
            # Alternative hypothesis: mean(vals_10) > mean(vals_01)
            t, p = stats.ttest_ind(vals_10, vals_01, equal_var=False, alternative='greater') # Welch’s t-test
        else:
            t, p = np.nan, np.nan
        test_results.append({"protein": prot, "t": t, "p": p,
                             "mean_01": vals_01.mean() if len(vals_01)>0 else np.nan,
                             "mean_10": vals_10.mean() if len(vals_10)>0 else np.nan})
        signif = "(*)" if p < 0.05 else ""
        ax.text(0.5, y_max - 0.04*(y_max - y_min), f"p = {p:.3f} {signif}",
                ha="center", va="top", fontsize=11, color="black")

        ax.set_title(prot.upper(),fontsize=15)
        ax.set_xlim(-0.5, 1.5)
        ax.set_xticks([0, 1], labels=[moi_labels[0], moi_labels[1]])
        ax.set_ylim(y_min, y_max)
        ax.grid(True, axis="y", alpha=0.3)

    ylabel = f"{title_additions} muts {'log10(' if use_log else ''}{metric if metric!='sum' else ''}{' )' if use_log else ''} freq {metric if metric=='sum' else ''}"
    for ax in axes[::2]:
        ax.set_ylabel(ylabel, fontsize=13)
#     fig.suptitle(f"Per-line {('log10 ' if use_log else '')}{metric} frequencies (Welch’s t-tests per protein), filtration: freq {freq_threshold}\n{title_additions}",
#                  y=0.98)

    handles = [plt.Line2D([0],[0], marker='o', color='none',
                          markerfacecolor=color_map[ln], markeredgecolor="black",
                          markersize=np.sqrt(point_size), label=str(ln))
               for ln in color_map]
    axes[1].legend(handles, list(color_map.keys()), title="line",
                   bbox_to_anchor=(1.05, 1.0), loc="upper left", borderaxespad=0.0)

    plt.tight_layout(rect=[0, 0, 0.88, 0.96])

    test_df = pd.DataFrame(test_results)
    print("\n=== Welch's t-tests per protein ===")
    print(test_df.to_string(index=False))
    return fig, axes, test_df
```

```python
def plot_shifted_moi_grid_with_brackets(
    syn_dfs, nonsyn_dfs, stop_dfs,
    proteins=("mat", "cp", "lys", "rep"),
    figsize=(12, 8),
    moi_labels=("MOI 0.1", "MOI 10"),
    point_size=60
):
    def _prep(df_tuple, mut_label, metric_col):
        d01, d10 = df_tuple
        d01 = d01.reset_index(); d01['MOI'] = moi_labels[0]
        d10 = d10.reset_index(); d10['MOI'] = moi_labels[1]
        combined = pd.concat([d01, d10], ignore_index=True)
        combined['mut_type'] = mut_label
        combined['val'] = combined[metric_col]
        return combined[['line', 'protein', 'MOI', 'mut_type', 'val']]

    # Data Prep
    df_all = pd.concat([
        _prep(syn_dfs, 'syn', 'mean_freq'),
        _prep(nonsyn_dfs, 'nonsyn', 'mean_freq'),
        _prep(stop_dfs, 'stop', 'sum_freq')
    ], ignore_index=True)
    df_all['protein'] = pd.Categorical(df_all['protein'].str.lower(), categories=proteins)
    
    # List to store statistical results
    stats_records = []

    # Global Y-limits 
    mean_vals = df_all[df_all['mut_type'].isin(['syn', 'nonsyn'])]['val']
    y_max_left = mean_vals.max() * 1.6 if not mean_vals.empty else 0.002
    
    sum_vals = df_all[df_all['mut_type'] == 'stop']['val']
    y_max_right = sum_vals.max() * 1.6 if not sum_vals.empty else 0.02

    # Initialize Figure
    fig, axes = plt.subplots(2, 2, figsize=figsize, sharey=True)
    axes = axes.flatten()
    
    styles = {
        'syn':    {'marker': 'o', 'color': '#1f77b4', 'label': 'Syn (mean)', 'offset': -0.18, 'axis': 'left'},
        'nonsyn': {'marker': '^', 'color': '#ff7f0e', 'label': 'Nonsyn (mean)', 'offset': 0.0, 'axis': 'left'},
        'stop':   {'marker': 'D', 'color': '#d62728', 'label': 'Nonsense (sum)', 'offset': 0.18, 'axis': 'right'}
    }
    x_base = {moi_labels[0]: 0, moi_labels[1]: 1}

    def draw_bracket(ax, x1, x2, y_start, y_top, text=None):
        """Draws a bracket that stays strictly above y_start."""
        # Square bracket: (x1, y_start) -> (x1, y_top) -> (x2, y_top) -> (x2, y_start)
        ax.plot([x1, x1, x2, x2], [y_start, y_top, y_top, y_start], 
                color='black', lw=1, solid_capstyle='butt', zorder=4)
        if text:
            ax.text((x1 + x2) / 2, y_top + (y_top * 0.005), text, 
                    ha='center', va='bottom', fontsize=11, fontweight='bold', zorder=5)

    # Plotting
    right_axes = [] 
    for i, (ax, prot) in enumerate(zip(axes, proteins)):
        ax_right = ax.twinx()
        right_axes.append(ax_right)
        sub = df_all[df_all['protein'] == prot]
        
        levels = {'syn': 0.78, 'nonsyn': 0.88, 'stop': 0.68} 

        for m_type in ['syn', 'nonsyn', 'stop']:
            m_sub = sub[sub['mut_type'] == m_type]
            style = styles[m_type]
            target_ax = ax if style['axis'] == 'left' else ax_right
            offset = style['offset']
            
            # --- Plot Data ---
            for ln, grp in m_sub.groupby('line'):
                vals = grp.set_index('MOI')['val']
                if all(m in vals.index for m in moi_labels):
                    x_coords = [0 + offset, 1 + offset]
                    y_coords = [vals[moi_labels[0]], vals[moi_labels[1]]]
                    target_ax.plot(x_coords, y_coords, color=style['color'], alpha=0.15, lw=1)
                
                for moi in moi_labels:
                    if moi in vals.index:
                        target_ax.scatter(x_base[moi] + offset, vals[moi], 
                                          marker=style['marker'], color=style['color'], 
                                          edgecolor='black', s=point_size, zorder=3, alpha=0.75)

            # --- Statistical Brackets ---
            v01 = m_sub[m_sub['MOI'] == moi_labels[0]]['val'].dropna()
            v10 = m_sub[m_sub['MOI'] == moi_labels[1]]['val'].dropna()
            
            if len(v01) >= 2 and len(v10) >= 2:
                t_stat, p = stats.ttest_ind(v10, v01, equal_var=False, alternative='greater')
                stats_records.append({
                    'protein': prot,
                    'mut_type': m_type,
                    't_stat': t_stat,
                    'p_value': p,
                    'mean_0.1': v01.mean(),
                    'mean_10': v10.mean()
                })
                y_limit = y_max_left if style['axis'] == 'left' else y_max_right
                
                # The 'legs' start just above the highest point in each column
                leg_y = max(v01.max(), v10.max()) + (y_limit * 0.02)
                bar_y = y_limit * levels[m_type]
                
                if bar_y <= leg_y: bar_y = leg_y + (y_limit * 0.05)
                

                if p<0.001: txt = "(***)"
                elif p<0.01: txt = "(**)"
                elif p < 0.05: txt = "(*)"
                else: txt = None
                    
                draw_bracket(target_ax, 0 + offset, 1 + offset, leg_y, bar_y, text=txt)

        # Formatting
        ax.set_ylim(0, y_max_left)
        ax_right.set_ylim(0, y_max_right)
        ax.set_title(prot.upper(), fontweight='bold', pad=15, fontsize=18)
        ax.set_xticks([0, 1]); ax.set_xticklabels(moi_labels, fontsize=14)
        ax.set_xlim(-0.6, 1.6)
        ax.grid(True, axis='y', alpha=0.1)

        if i % 2 == 0: 
            ax.set_ylabel('Mean Mutation Freq', fontsize=16)
        else: 
            ax.tick_params(labelleft=False)
        
        if i in [1, 3]: 
            ax_right.set_ylabel('Sum Freq (Nonsense)', fontsize=16, labelpad=15)
        else: 
            ax_right.tick_params(labelright=False)

    for rax in right_axes: rax.set_ylim(0, y_max_right)

    legend_elements = [Line2D([0], [0], marker=v['marker'], color='w', label=v['label'],
                              markerfacecolor=v['color'], markeredgecolor='k', markersize=12)
                       for v in styles.values()]
    fig.legend(handles=legend_elements, loc='lower center', bbox_to_anchor=(0.38, 0.54),
               ncol=1, frameon=True, fontsize=14) 
    fig.set_dpi(300)
    plt.tight_layout()
    df_stats = pd.DataFrame(stats_records)
    return fig, df_stats
```

### plot p-value of mean freq difference as a function of threshold

```python
def get_p_values_for_threshold(freqs_df_01, freqs_df_10, threshold, metric='mean', use_log=True, proteins=("mat", "cp", "lys", "rep")):
    summary_01 = create_summary_ferq_df(freqs_df_01, threshold, greater_than=False)
    summary_10 = create_summary_ferq_df(freqs_df_10, threshold, greater_than=False)
    
    col = f"log10_{metric}" if use_log else f"{metric}_freq"
    results = {}

    for prot in proteins:
        try:
            vals_01 = summary_01.xs(prot, level='protein')[col].dropna()
            vals_10 = summary_10.xs(prot, level='protein')[col].dropna()
            
            if len(vals_01) >= 2 and len(vals_10) >= 2:
                # If values are nearly identical, add a negligible amount of noise
                if vals_01.std() < 1e-12:
                    vals_01 = vals_01 + np.random.normal(0, 1e-12, len(vals_01))
                if vals_10.std() < 1e-12:
                    vals_10 = vals_10 + np.random.normal(0, 1e-12, len(vals_10))

                t, p = stats.ttest_ind(vals_10, vals_01, equal_var=False, alternative='greater')
            else:
                p = np.nan
        except (KeyError, ZeroDivisionError):
            p = np.nan
            
        results[prot] = p
        
    return results
```

```python
def plot_merged_p_value_vs_threshold(df_01, df_10, thresholds, metric='mean', use_log=True):
    """ 
    Nonsyn = solid lines, Syn = dashed lines.
    """
    proteins = ("mat", "cp", "lys", "rep")
    colors = ['peru', 'seagreen', 'indianred', 'darkturquoise']
    
    plt.figure(figsize=(13, 8))
    
    for mut_type, lstyle in [('nonsyn', '-'), ('syn', '--')]:
        raw_p_values = {prot: [] for prot in proteins}
        
        sub_01 = df_01[df_01['syn_nonsyn'] == mut_type]
        sub_10 = df_10[df_10['syn_nonsyn'] == mut_type]
        
        for t in thresholds:
            p_dict = get_p_values_for_threshold(sub_01, sub_10, t, metric, use_log, proteins)
            for prot in proteins:
                raw_p_values[prot].append(p_dict[prot])
        
        # FDR Correction for the current mutation type
        flat_p = []
        indices = []
        for prot in proteins:
            for i, val in enumerate(raw_p_values[prot]):
                if not np.isnan(val):
                    flat_p.append(val)
                    indices.append((prot, i))

        if len(flat_p) > 0:
            _, corrected_p, _, _ = multipletests(flat_p, method='fdr_bh')
            
            adj_p_dict = {prot: [np.nan] * len(thresholds) for prot in proteins}
            for (prot, i), adj_p in zip(indices, corrected_p):
                adj_p_dict[prot][i] = adj_p
            
            for i, prot in enumerate(proteins):
                label = f"{prot} {mut_type}"
                plt.plot(thresholds, adj_p_dict[prot], color=colors[i], 
                         linestyle=lstyle, marker='o', markersize=4, 
                         linewidth=2, label=label, alpha=0.8)


    plt.axhline(y=0.05, color='black', linestyle=':', alpha=0.7, label='Alpha = 0.05')
    plt.yscale('log')
    plt.gca().invert_yaxis() 
    
    plt.xlabel("Frequency Threshold", fontsize=16, labelpad=10)
    plt.ylabel("Adjusted P-value (Log Scale)", fontsize=16, labelpad=10)
    
#     # Create a custom legend to differentiate Mutation Types and Proteins
#     plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=10)
    plt.grid(True, which="both", ls="-", alpha=0.15)
#     plt.tight_layout()
#     plt.show()
    
    protein_handles = [Line2D([0], [0], color=colors[i], lw=2, label=prot) 
                       for i, prot in enumerate(proteins)]
    
    # mutation styles
    style_handles = [
        Line2D([0], [0], color='gray', linestyle='-', lw=2, label='nonsyn'),
        Line2D([0], [0], color='gray', linestyle='--', lw=2, label='syn'),
        Line2D([0], [0], color='black', linestyle=':', lw=1, label='alpha = 0.05')
    ]

    first_legend = plt.legend(handles=protein_handles, title="Proteins",
                              bbox_to_anchor=(1.02, 1), loc='upper left', frameon=True, fontsize=14,title_fontsize=14)
    first_legend.get_title().set_weight('bold')
    plt.gca().add_artist(first_legend)
    
    plt.legend(handles=style_handles, 
               bbox_to_anchor=(1.02, 0.75), loc='upper left', frameon=True, fontsize=14)
    plt.gcf().set_dpi(300)

    plt.tight_layout()
    plt.show()
```

```python
def plot_ms2_colored_replicates(df, window_size=50, step=20):
    genome_length = df['position'].max()
    windows = np.arange(0, genome_length - window_size, step)
    replicates = sorted(df['line'].unique())
    
    # filter out adaptive mutations 
    adaptive_col = df.columns[df.columns.str.contains('adaptive')][0]
    df = df[(df[adaptive_col]=='nonada')&(df['passage']==10)]
    df = df.drop_duplicates(subset=['mut_name', 'position','mutation','line','passage'])
    
    
    colors = plt.cm.tab10(np.linspace(0, 1, len(replicates)))
    
    rep_results_ns = {rep: [] for rep in replicates}
    rep_results_syn = {rep: [] for rep in replicates}
    x_coords = []

    for start in windows:
        mid = start + (window_size / 2)
        x_coords.append(mid)
        
        for rep in replicates:
            win_data = df[(df['line'] == rep) & 
                          (df['position'] >= start) & 
                          (df['position'] < (start + window_size))]
            
            ns_val = win_data[win_data['syn_nonsyn'] == 'nonsyn']['freq'].sum() / window_size
            syn_val = win_data[win_data['syn_nonsyn'] == 'syn']['freq'].sum() / window_size
            
            rep_results_ns[rep].append(ns_val)
            rep_results_syn[rep].append(syn_val)
            

    mean_ns = np.mean([rep_results_ns[r] for r in replicates], axis=0)
    mean_syn = np.mean([rep_results_syn[r] for r in replicates], axis=0)

    fig, ax = plt.subplots(figsize=(16, 8))
    #  Plot Synonymous Control
    for rep in replicates:
        ax.plot(x_coords, rep_results_syn[rep], color='gray', alpha=0.1, linewidth=0.8)
    ax.plot(x_coords, mean_syn, color='gray', linewidth=1.5, linestyle='--', label='Mean Synonymous')

    # Plot Non-synonymous Lines
    for i, rep in enumerate(replicates):
        ax.plot(x_coords, rep_results_ns[rep], color=colors[i], alpha=0.3, linewidth=1.2, label=f'line {rep} (Non-synonymous)')
    
    ax.plot(x_coords, mean_ns, color='black', linewidth=2.3, label='Mean Non-synonymous', zorder=5)

    proteins = [
        {"name": "Maturation protein", "start": 130, "end": 1311},
        {"name": "Coat", "start": 1335, "end": 1727},
        {"name": "Lysis", "start": 1678, "end": 1905}, 
        {"name": "Replicase", "start": 1761, "end": 3398}
    ]
    
    y_max = max(np.max(mean_ns), np.max(mean_syn))
    base_y = y_max * 1.25
    
    for p in proteins:
        y_pos = base_y * 1.15 if p['name'] == "Lysis" else base_y
        ax.hlines(y_pos, p['start'], p['end'], colors='black', lw=2)
        ax.vlines([p['start'], p['end']], y_pos*0.97, y_pos*1.03, colors='black', lw=1.5)
        ax.text((p['start'] + p['end'])/2, y_pos * 1.05, p['name'], 
                ha='center', va='bottom', fontweight='bold', fontsize=14)

    ax.set_ylim(0, base_y * 1.4)
    ax.set_xlabel("Genome Position", fontsize=18,labelpad=15)
    ax.set_ylabel("Mutation Frequency", fontsize=18,labelpad=15) #  / Window Size
    
    handles, labels = ax.get_legend_handles_labels()
    order = [i for i, l in enumerate(labels) if 'line' in l] + \
            [labels.index('Mean Non-synonymous'), labels.index('Mean Synonymous')]
    ax.legend([handles[idx] for idx in order], [labels[idx] for idx in order], 
              loc='upper right',fontsize=12, frameon=True, ncol=2)
    
    ax.xaxis.set_major_locator(MultipleLocator(250))

    ax.xaxis.set_minor_locator(MultipleLocator(50))

    ax.grid(axis='x', which='both', alpha=0.1)

    plt.tight_layout()
    plt.show()
```

```python
def load_rate4site_data(file_path):
    """Parses the Rate4Site .res file and returns a cleaned DataFrame."""
    data = []
    with open(file_path, 'r') as f:
        for line in f:
            if line.startswith('#') or not line.strip():
                continue
            parts = line.split()
            # Expected columns: POS, SEQ, SCORE, QQ-INTERVAL, STD, MSA DATA
            try:
                data.append([int(parts[0]), parts[1], float(parts[2])])
            except (ValueError, IndexError):
                continue
    
    return pd.DataFrame(data, columns=['POS', 'NT', 'SCORE'])

def add_gene_annotations(ax, genes):
    """Adds horizontal bars and labels for each gene region."""
    colors = ['#2ecc71', '#3498db', '#e67e22', '#9b59b6']
    
    ymin, ymax = ax.get_ylim()
    annotation_y = ymin - (ymax - ymin) * 0.1  # Place slightly below the plot
    
    for i, (gene, (start, end)) in enumerate(genes.items()):
        color = colors[i % len(colors)]
        ax.hlines(y=annotation_y - (i*0.2), xmin=start, xmax=end, 
                  color=color, linewidth=6, label=gene)
        ax.text((start + end) / 2, annotation_y - (i*0.2) + 0.05, 
                gene, ha='center', va='bottom', fontsize=9, fontweight='bold', color=color)

def plot_genomic_conservation(df, genes, output_file='ms2_genome_conservation.png'):
    """Creates the visualization for genomic conservation."""
    plt.figure(figsize=(20, 8))
    ax = plt.gca()

    plt.plot(df['POS'], df['SCORE'], color='black', linewidth=0.5, alpha=0.6)

    plt.fill_between(df['POS'], df['SCORE'], 0, where=(df['SCORE'] < 0), 
                     color='royalblue', alpha=0.5, label='Conserved')
    plt.fill_between(df['POS'], df['SCORE'], 0, where=(df['SCORE'] > 0), 
                     color='crimson', alpha=0.5, label='Variable')

    plt.axhline(0, color='black', linestyle='-', linewidth=0.8, alpha=0.3)
    ax.xaxis.set_major_locator(ticker.MultipleLocator(250))
    ax.xaxis.set_minor_locator(ticker.MultipleLocator(50))
    
    plt.title('MS2 Full Genome Nucleotide Conservation Profile (Rate4Site)', fontsize=18)
    plt.xlabel('Nucleotide Position', fontsize=14)
    plt.ylabel('Conservation Score', fontsize=14)
    
    add_gene_annotations(ax, genes)
    
    plt.grid(axis='y', linestyle=':', alpha=0.5)
    plt.tight_layout()
    plt.savefig(output_file, dpi=300)
    plt.show()

def plot_genomic_conservation_sliding_window(df, genes, window_size=100, output_file='ms2_genome_conservation_smoothed.png'):
    """
    Creates a visualization for genomic conservation using a sliding window average.
    """

    df_smoothed = df.copy()
    df_smoothed['SMOOTH_SCORE'] = df['SCORE'].rolling(window=window_size, center=True).mean()

    plt.figure(figsize=(20, 8))
    ax = plt.gca()

    # Drop NaNs created at the edges by the rolling window for a clean plot
    plot_data = df_smoothed.dropna(subset=['SMOOTH_SCORE'])

    plt.plot(plot_data['POS'], plot_data['SMOOTH_SCORE'], color='black', linewidth=1.5, alpha=0.8)

    plt.fill_between(plot_data['POS'], plot_data['SMOOTH_SCORE'], 0, 
                     where=(plot_data['SMOOTH_SCORE'] < 0), 
                     color='royalblue', alpha=0.4, label='Conserved (Window)')
    plt.fill_between(plot_data['POS'], plot_data['SMOOTH_SCORE'], 0, 
                     where=(plot_data['SMOOTH_SCORE'] > 0), 
                     color='crimson', alpha=0.4, label='Variable (Window)')

    plt.axhline(0, color='black', linestyle='--', linewidth=1, alpha=0.3)
    ax.xaxis.set_major_locator(ticker.MultipleLocator(250))
    ax.xaxis.set_minor_locator(ticker.MultipleLocator(50))
    
    plt.title(f'MS2 Full Genome Conservation Profile (Sliding Window: {window_size}nt)', fontsize=18)
    plt.xlabel('Nucleotide Position', fontsize=14)
    plt.ylabel(f'Mean Conservation Score (n={window_size})', fontsize=14)
    
    add_gene_annotations(ax, genes)
    
    plt.grid(axis='y', linestyle=':', alpha=0.5)
    plt.tight_layout()
    plt.savefig(output_file, dpi=300)
    plt.show()
    return plot_data
```

```python
def filter_and_map_conservation(empirical_df, rate4site_df, passage=10):
    """
    Filters for nonsynonymous mutations at a specific passage and 
    maps Rate4Site conservation scores to each mutation.
    """
    adaptive_cols = [col for col in empirical_df.columns if 'adaptive_moi' in col]
    adaptive_col = adaptive_cols[0] if adaptive_cols else None
    filtered_df = empirical_df[
        (empirical_df['passage'] == passage) & 
        (empirical_df['syn_nonsyn'] == 'nonsyn') &
        (empirical_df[adaptive_col] == 'nonada')
    ].copy()  # 
    
    # Merge with Rate4Site data based on position
    # empirical 'position' matches Rate4Site 'POS'
    merged_df = pd.merge(
        filtered_df, 
        rate4site_df[['POS','SCORE', 'SMOOTH_SCORE']], 
        left_on='position', 
        right_on='POS', 
        how='left'
    )

    merged_df = merged_df.drop(columns=['POS']).rename(columns={'SCORE': 'cons_score', 'SMOOTH_SCORE':'avg_cons_score'})
    
    return merged_df

def process_all_moi_data(df_moi10, df_moi01, rate4site_df):
    """Processes both MOI datasets and returns them with conservation scores."""
    
    print("Mapping scores for MOI 10...")
    scored_moi10 = filter_and_map_conservation(df_moi10, rate4site_df)
    
    print("Mapping scores for MOI 0.1...")
    scored_moi01 = filter_and_map_conservation(df_moi01, rate4site_df)
    
    return scored_moi10, scored_moi01

scored_moi10, scored_moi01 = process_all_moi_data(moi10_freqs_linesEGH, moi01_freqs, rate4site_df)

print("\nMOI 10 Nonsyn Mutations with Scores:")
print(scored_moi10[['position', 'protein', 'freq', 'cons_score','avg_cons_score']].head())
```

```python
def plot_normalized_conservation(scored_moi10, scored_moi01, full_genome_scores, genes, avg=False):
    """
    Plots mutation conservation distributions against the 'Background' 
    conservation profile of each gene.
    """
    
    score = 'avg_cons_score' if avg else 'cons_score'
    proteins = [p for p in genes.keys()]
    fig, axes = plt.subplots(len(proteins), 1, figsize=(12, 5 * len(proteins)), sharex=True)

    for ax, protein in zip(axes, proteins):
        start, end = genes[protein]
        
        gene_background = full_genome_scores[
            (full_genome_scores['POS'] >= start) & (full_genome_scores['POS'] <= end)
        ]['SCORE']


        obs_10 = scored_moi10[scored_moi10['protein'].str.lower() == protein.lower()][score]
        obs_01 = scored_moi01[scored_moi01['protein'].str.lower() == protein.lower()][score]

        # Plotting the Background (The 'Expected' distribution)
        sns.kdeplot(gene_background, ax=ax, color='gray', fill=True, alpha=0.1, 
                    label='Gene Background (All Sites)', linewidth=1, linestyle='--')

        # Plotting the Observed Mutations
        sns.kdeplot(obs_10, ax=ax, color='#e74c3c', label='MOI 10 (Observed)', linewidth=2.5)
        sns.kdeplot(obs_01, ax=ax, color='#3498db', label='MOI 0.1 (Observed)', linewidth=2.5)

        ax.set_title(f'Normalized Conservation Enrichment: {protein}', fontsize=15, fontweight='bold')
        ax.set_ylabel('Density')
        ax.legend(loc='upper left')
        ax.grid(axis='y', linestyle=':', alpha=0.5)

    plt.xlabel('Rate4Site Conservation Score')
    plt.tight_layout()
    # plt.savefig('normalized_conservation_enrichment.png', dpi=300)
    plt.show()


plot_normalized_conservation(scored_moi10, scored_moi01, rate4site_df, GENE_MAP)
```

```python
def compare_moi_distributions(scored_moi10, scored_moi01, genes, count_threshold=0):
    """
    Compares conservation score distributions between MOI 10 and MOI 0.1 
    for each protein.
    """
    comparison_results = []
    
    
    df10 = scored_moi10[scored_moi10['read_count']>count_threshold].copy()
    df01 = scored_moi01[scored_moi01['read_count']>count_threshold].copy()


    for protein in genes.keys():
        scores_10 = df10[df10['protein'] == protein]['cons_score'].dropna()
        scores_01 = df01[df01['protein'] == protein]['cons_score'].dropna()
        
        if len(scores_10) > 2 and len(scores_01) > 2:
            m10_med, m01_med = scores_10.median(), scores_01.median()
            m10_avg, m01_avg = scores_10.mean(), scores_01.mean()
            
            median_shift = m10_med - m01_med
            mean_shift = m10_avg - m01_avg
            
            # Mann-Whitney U Test 
            stat, p_val = mannwhitneyu(scores_10, scores_01, alternative='less')
            
            comparison_results.append({
                'Protein': protein,
                'n_MOI_10': len(scores_10),
                'n_MOI_0.1': len(scores_01),
                'Median_MOI_10': round(m10_med, 3),
                'Median_MOI_0.1': round(m01_med, 3),
                'Median_Shift': round(median_shift, 3),
                'Mean_MOI_10': round(m10_avg, 3),
                'Mean_MOI_0.1': round(m01_avg, 3),
                'Mean_Shift': round(mean_shift, 3),
                'p_value': f"{p_val:.4e}"
            })
            
    return pd.DataFrame(comparison_results)
```

```python
comparison_df = compare_moi_distributions(scored_moi10, scored_moi01, GENE_MAP,3)

print("--- Statistical Comparison: MOI 10 vs. MOI 0.1 --- (read_count>3)")
print(comparison_df.to_string(index=False))
```

```python
def get_reg_params(df):
    """Calculates regression parameters for the given DataFrame."""
    x = df['passage']
    y = df['freq']
    slope, intercept, r_val, p_val, std_err = stats.linregress(x, y)
    return slope, intercept, r_val**2

def plot_moi_shifted_boxplot(df_moi01, df_moi10):
    # 1. Calculate regression stats
    slope01, intercept01, r2_01 = get_reg_params(df_moi01)
    slope10, intercept10, r2_10 = get_reg_params(df_moi10)

    fig, ax = plt.subplots(figsize=(10, 6))

    def get_bp_data(df):
        passages = sorted(df['passage'].unique())
        return [df[df['passage'] == p]['freq'] for p in passages], np.array(passages)

    data01, pass01 = get_bp_data(df_moi01)
    data10, pass10 = get_bp_data(df_moi10)

    offset = 0.05
    pos01 = pass01 - offset
    pos10 = pass10 + offset

    def get_style(color):
        return {
            "patch_artist": True,
            "manage_ticks": False,
            "widths": 0.4,
            "boxprops": dict(facecolor=color, alpha=0.6, color=color, edgecolor='black', linewidth=1.5),
            "medianprops": dict(color=color, linewidth=2),
            "whiskerprops": dict(color='black'),
            "capprops": dict(color='black'),
            "flierprops": dict(marker='o', markerfacecolor=color, markeredgecolor=color, alpha=0.5)
        }

    ax.boxplot(data01, positions=pos01, **get_style('#2c7bb6'))
    ax.boxplot(data10, positions=pos10, **get_style('#d7191c'))

    x_range = np.array([0, 10])
    ax.plot(x_range, slope01 * x_range + intercept01, 
            color='#2c7bb6', linestyle='--', linewidth=2)
    ax.plot(x_range, slope10 * x_range + intercept10, 
            color='#d7191c', linestyle='--', linewidth=2)

    ax.set_xlabel("passage", fontsize=18, labelpad=15)
    ax.set_ylabel("syn frequency", fontsize=18, labelpad=15)
    
    ax.set_xlim(-0.5, 11)
    ax.set_ylim(0, 0.0035)
    ax.set_xticks([0, 3, 5, 7, 8, 10])
    ax.grid(True, axis='y', linestyle='--', alpha=0.3)

    text_str = (f"MOI 0.1 Slope: {slope01:.2e}\n"
                f"MOI 10.0 Slope: {slope10:.2e}")
    ax.text(0.02, 0.98, text_str, transform=ax.transAxes, fontsize=14,
            verticalalignment='top', bbox=dict(boxstyle='round', facecolor='None', alpha=0.8))

    legend_elements = [Line2D([0], [0], color='#2c7bb6', lw=2, label='MOI 0.1'),
                       Line2D([0], [0], color='#d7191c', lw=2, label='MOI 10')]
    ax.legend(handles=legend_elements, loc='upper left', bbox_to_anchor=(0.001, 0.88), fontsize=16)
    
    plt.gcf().set_dpi(300)
    plt.tight_layout()
    plt.show()

```
