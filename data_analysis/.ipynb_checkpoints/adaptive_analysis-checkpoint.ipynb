```python
def top_mutations_on_adaptive_genomes(df, adaptive_muts, top_n=10):
    def parse_list(x):
        if isinstance(x, list):
            return x
        if pd.isna(x):
            return []
        if isinstance(x, str):
            try:
                return ast.literal_eval(x)
            except Exception:
                return x.split('_')
        return []

    df = df.copy()
    df['mut_name_parsed'] = df['mut_name'].apply(parse_list)
    df['protein_parsed'] = df['protein'].apply(parse_list)

    # one row per mutation-protein pair per genome
    exploded = df.explode(['mut_name_parsed', 'protein_parsed'])

    # remove mutations that define adaptive genomes
    exploded = exploded[
        ~exploded['mut_name_parsed'].isin(adaptive_muts)
    ]

    # count appearances on adaptive / non-adaptive genomes, per line
    counts = (
        exploded
        .groupby(['line', 'mut_name_parsed', 'protein_parsed', 'adaptive_genome'])
        .size()
        .reset_index(name='count')
    )

    counts_wide = (
        counts
        .pivot_table(
            index=['line', 'mut_name_parsed', 'protein_parsed'],
            columns='adaptive_genome',
            values='count',
            fill_value=0
        )
        .reset_index()
        .rename(columns={
            True: 'count_on_adaptive_genomes',
            False: 'count_on_nonadaptive_genomes'
        })
    )

    # make sure both columns exist
    for col in ['count_on_adaptive_genomes', 'count_on_nonadaptive_genomes']:
        if col not in counts_wide.columns:
            counts_wide[col] = 0

    # number of adaptive genomes per line
    n_adaptive = (
        df[df['adaptive_genome'] == True]
        .groupby('line')
        .size()
        .rename('n_adaptive_genomes')
        .reset_index()
    )

    counts_wide = counts_wide.merge(n_adaptive, on='line', how='left')

    counts_wide['prob_on_adaptive_genomes'] = (
        counts_wide['count_on_adaptive_genomes'] /
        counts_wide['n_adaptive_genomes']
    )

    # probability that an appearance of this mutation is on an adaptive genome
    counts_wide['prob_adaptive_given_mutation'] = (
        counts_wide['count_on_adaptive_genomes'] /
        (
            counts_wide['count_on_adaptive_genomes'] +
            counts_wide['count_on_nonadaptive_genomes']
        )
    )

    top_mutations = (
        counts_wide
        .sort_values(
            ['line', 'prob_on_adaptive_genomes', 'count_on_adaptive_genomes'],
            ascending=[True, False, False]
        )
        .groupby('line')
        .head(top_n)
        .reset_index(drop=True)
    )

    return top_mutations
```

```python
def summarize_short_read_appearance(
    short_df,
    mut_list_df,
    min_freq=0,
    keep_all_short_read_lines=True
):
    """
    For each candidate mutation from long-read data, find where it appears
    in the short-read dataframe.

    Parameters
    ----------
    short_df : pd.DataFrame
        Short-read mutation dataframe.
        Required columns:
        mut_name, freq, passage, line

    mut_list_df : pd.DataFrame
        Candidate mutation dataframe from long reads.
        Required columns:
        long_read_line, mut_name
        Optional columns:
        long_read_freq, protein, syn_nonsyn

    min_freq : float
        Minimum short-read frequency to count as appearing.
        Use 0 to include every positive frequency.

    keep_all_short_read_lines : bool
        If True, search for each mutation in all short-read lines.
        If False, search only in the original long-read line.

    Returns
    -------
    detailed_df : pd.DataFrame
        One row per candidate mutation observed in short reads.

    summary_df : pd.DataFrame
        One row per candidate mutation, including mutations not observed
        in short reads.
    """

    required_short_cols = {"mut_name", "freq", "passage", "line"}
    missing_short = required_short_cols - set(short_df.columns)
    if missing_short:
        raise ValueError(f"short_df is missing required columns: {missing_short}")

    required_candidate_cols = {"long_read_line", "mut_name"}
    missing_candidates = required_candidate_cols - set(mut_list_df.columns)
    if missing_candidates:
        raise ValueError(f"mut_list_df is missing required columns: {missing_candidates}")

    df = short_df.copy()
    candidates = mut_list_df.copy()

    # Standardize types
    df["mut_name"] = df["mut_name"].astype(str)
    df["line"] = df["line"].astype(str)
    df["passage"] = pd.to_numeric(df["passage"], errors="coerce")
    df["freq"] = pd.to_numeric(df["freq"], errors="coerce")

    candidates["mut_name"] = candidates["mut_name"].astype(str)
    candidates["long_read_line"] = candidates["long_read_line"].astype(str)

    # Keep one row per candidate mutation
    candidates = candidates.drop_duplicates(
        subset=["long_read_line", "mut_name"]
    ).reset_index(drop=True)

    # Avoid duplicated annotation columns.
    # We keep protein/syn_nonsyn from candidates in the final summary.
    # In detailed_df, short-read columns are kept, and only candidate-specific
    # columns that do not already exist in short_df are added.
    candidate_cols_to_add = [
        col for col in candidates.columns
        if col == "mut_name" or col not in df.columns
    ]

    detailed = df.merge(
        candidates[candidate_cols_to_add],
        on="mut_name",
        how="inner"
    )

    if not keep_all_short_read_lines:
        detailed = detailed[detailed["line"] == detailed["long_read_line"]]

    detailed = detailed[detailed["freq"] > min_freq].copy()

    detailed = detailed.sort_values(
        ["long_read_line", "mut_name", "line", "passage"]
    ).reset_index(drop=True)

    def summarize_group(g):
        g_sorted = g.sort_values(["line", "passage"])

        lines = sorted(g_sorted["line"].dropna().unique())
        passages = sorted(g_sorted["passage"].dropna().astype(int).unique())

        line_passages = (
            g_sorted
            .groupby("line")["passage"]
            .apply(lambda x: sorted(x.dropna().astype(int).unique()))
            .to_dict()
        )

        freqs_by_line_passage = (
            g_sorted
            .groupby(["line", "passage"])["freq"]
            .max()
            .reset_index()
            .sort_values(["line", "passage"])
        )

        freq_trace = {
            line: dict(zip(sub["passage"].astype(int), sub["freq"]))
            for line, sub in freqs_by_line_passage.groupby("line")
        }

        first_passage = g_sorted["passage"].min()
        max_freq = g_sorted["freq"].max()
        max_row = g_sorted.loc[g_sorted["freq"].idxmax()]

        return pd.Series({
            "short_read_lines_seen": lines,
            "short_read_passages_seen": passages,
            "passages_by_line": line_passages,
            "freq_trace_by_line": freq_trace,
            "first_short_read_passage_seen": (
                int(first_passage) if pd.notna(first_passage) else np.nan
            ),
            "max_short_read_freq": max_freq,
            "line_of_max_short_read_freq": max_row["line"],
            "passage_of_max_short_read_freq": int(max_row["passage"]),
        })

    if len(detailed) > 0:
        summary = (
            detailed
            .groupby(["long_read_line", "mut_name"], as_index=False)
            .apply(summarize_group) # , include_groups=False
            .reset_index(drop=True)
        )
    else:
        summary = pd.DataFrame(columns=["long_read_line", "mut_name"])

    # Final merge keeps ALL candidate mutations, including those absent in short reads.
    # Candidate annotations such as protein/syn_nonsyn/long_read_freq stay clean.
    summary = candidates.merge(
        summary,
        on=["long_read_line", "mut_name"],
        how="left"
    )

    summary["seen_in_short_reads"] = summary["max_short_read_freq"].notna()

    return detailed, summary
```
